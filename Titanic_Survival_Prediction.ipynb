```python
# Part 3: Data Preprocessing

# Check Missing Values
df.isnull().sum()
```

```python
# Fill Missing Values
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
# Verify Missing Values
df.isnull().sum()
```

```python
# Label Encoding / One-Hot Encoding
print("No Label Encoding or One-Hot Encoding is required because the categorical features are already encoded.")
```

```python
# Separate Features and Target Variable
X = df.drop(["Passengerid", "Survived"], axis=1)
y = df["Survived"]
print(X.head())
print(y.head())
```

```python
# Feature Scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X_scaled,y,test_size=0.20,random_state=42)
print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)
```

# Data Preprocessing  
Before the training of the Logistic Regression algorithm, the dataset had to go through preprocessing.  

# Steps Performed
1. The missing values in column Embarked were filled in with the most frequent value, thus the mode.  
2. No Label Encoding and One-Hot Encoding were performed as the dataset already contains categorical values encoded.  
3. The variable Passengerid was deleted due to the fact that it only serves as a unique reference and does not play any role in prediction.  
4. Feature scaling was carried out by means of using the StandardScaler so that all the features are scaled similarly, which will raise Logistic Regression performance.  
5. The dataset was divided into 80% of training set and 20% of testing set.  

# Why These Steps Are Necessary  
- Missing Value Treatment: Helps to avoid problems while training the model as well as to use all possible information.  
- Encoding: Changes categorical information into its numerical equivalent. (Already done in this case.)  
- Feature Scaling: Makes sure that features with high values are not dominant in the fitting process.  
- Train-Test Splitting: Allows the model to be trained on part of the dataset and evaluated via other parts.  

```python
# Part 4: Logistic Regression Model

# Train Logistic Regression Model
model = LogisticRegression(random_state=42)
model.fit(X_train, y_train)
```

```python
# Model Coefficients
coefficients = pd.DataFrame({"Feature": X.columns,"Coefficient": model.coef_[0]})
coefficients
```

```python
# Model Intercept
print("Intercept:", model.intercept_[0])
```

```python
# Feature Importance
feature_importance = coefficients.copy()
feature_importance["Absolute Coefficient"] = feature_importance["Coefficient"].abs()
feature_importance = feature_importance.sort_values(by="Absolute Coefficient", ascending=False)
feature_importance
```

```python
# Feature Importance Visualization
plt.figure(figsize=(8,5))
sns.barplot(data=feature_importance,x="Absolute Coefficient",y="Feature")
plt.title("Feature Importance Based on Logistic Regression")
plt.xlabel("Absolute Coefficient")
plt.ylabel("Feature")
plt.show()
```

# Logistic Regression Model
Training of a Logistic Regression model was conducted using the training dataset to make predictions regarding the survival of the passengers.

# Observations
1. The Logistic Regression model has been successfully trained using the Scikit-Learn library.
2. The model's intercept value is -1.3446, meaning that there is a baseline prediction when all the feature values are zero.
3. Among the features, Sex has the greatest positive coefficient value (0.9053), which means that it has the greatest effect regarding the positive side of survival prediction.
4. Pclass, on the other hand, has the highest negative coefficient value (-0.6349), illustrating that the passenger class has a substantial influence on his/her survival.
5. A negative coefficient value of -0.3365 for Age means that as the age increases, so does the chance of surviving the travel, though very slight.
6. Other features like sibsp, Embarked, and Parch have smaller negative coefficient values, meaning their influence on prediction is less than that of other variables.
7. Regarding Fare, it has a coefficient value close to zero (-0.0038), meaning it shall have a weak influence on the prediction.
8. Overall, based on coefficients' absolute values, the most influential variables are Sex, Pclass, and Age.

```python
# Part 5: Model Evaluation

# Predict on Test Data
y_pred = model.predict(X_test)
```

```python
# Accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
# Precision
precision = precision_score(y_test, y_pred)
print("Precision:", precision)
```

```python
# Recall
recall = recall_score(y_test, y_pred)
print("Recall:", recall)
```

```python
# F1 Score
f1 = f1_score(y_test, y_pred)
print("F1 Score:", f1)
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
# Confusion Matrix Heatmap
plt.figure(figsize=(6,5))
sns.heatmap(cm,annot=True,fmt="d",cmap="Blues")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Confusion Matrix")
plt.show()
```

```python
# Classification Report
print(classification_report(y_test, y_pred))
```

# Model Evaluation
The Logistic Regression model was evaluated using different performance metrics to measure its effectiveness in predicting passenger survival.

# Results
- Accuracy: 76.72%
- Precision: 64.29%
- Recall: 36.99%
- F1 Score: 46.96%

# Confusion Matrix
- True Positives (TP): 27

# Comments
1. The accuracy of the model is 76.72%, which suggests that it correctly identified most passengers.
2. As the precision of the model is 64.29%, it has predicted surviving passengers with an accuracy of around 64%.
3. The Model's recall is 36.99%, which is lower than expected, indicating that the model identified many survivors.
4. The F1 score, which is equal to 46.96%, suggests that there is moderate balance between precision and recall.
5. According to the confusion matrix, the model performs much better in predicting non-survivors (Class 0) than survivors (Class 1) due to the imbalance in the dataset.


# What Metric is More Important?
In this regard, Recall is important since it measures how many real survivors are well identified by the model.
Because of the imbalance in the dataset, the model missed lots of survivors (46 false negatives) and improving recall would help the model find more survivors .
Even though the accuracy is high, it would be misleading to rely on accuracy alone since the model performs much better on the majority class (non-survivors) compared to the minority class (survivors).
